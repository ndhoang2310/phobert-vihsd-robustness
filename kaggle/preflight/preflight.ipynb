# ViHSD — kiểm tra môi trường GPU
Notebook riêng tư xác nhận tài nguyên trước khi chạy baseline.

```python
import json, platform, subprocess, urllib.request
from pathlib import Path
import torch
print(subprocess.check_output(["nvidia-smi"], text=True))
assert torch.cuda.is_available(), "GPU is required"
r = {"python": platform.python_version(), "torch": torch.__version__, "cuda": torch.version.cuda, "gpu": torch.cuda.get_device_name(0), "gpu_count": torch.cuda.device_count()}
for url in ["https://huggingface.co/api/models/vinai/phobert-base", "https://raw.githubusercontent.com/sonlam1102/vihsd/70fe50c9c94145e3b9be5614a048cbd5b480fa54/data/vihsd.zip"]:
    with urllib.request.urlopen(url, timeout=60) as response:
        r[url.split('/')[2]] = response.status
print(json.dumps(r, indent=2))
Path("/kaggle/working/preflight.json").write_text(json.dumps(r, indent=2))
x = torch.randn(32, 32, device="cuda", requires_grad=True)
x.square().mean().backward()
print("GPU_FORWARD_BACKWARD_OK")
```
